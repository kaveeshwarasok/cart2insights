```python
import pandas as pd
import os
from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 120)
```

```python
# Get the project root directory
PROJECT_ROOT = Path.cwd().parent

# Raw data directory
RAW_DATA_PATH = PROJECT_ROOT / "data" / "raw"

print("Project Root:", PROJECT_ROOT)
print("Raw Data Path:", RAW_DATA_PATH)
```

```python
DATASETS = {
    "customers": "olist_customers_dataset.csv",
    "geolocation": "olist_geolocation_dataset.csv",
    "order_items": "olist_order_items_dataset.csv",
    "order_payments": "olist_order_payments_dataset.csv",
    "order_reviews": "olist_order_reviews_dataset.csv",
    "orders": "olist_orders_dataset.csv",
    "products": "olist_products_dataset.csv",
    "sellers": "olist_sellers_dataset.csv",
    "translation": "product_category_name_translation.csv"
}

print(f"Total datasets expected: {len(DATASETS)}")

for name, filename in DATASETS.items():
    print(f"{name:20} -> {filename}")
```

```python
data = {}

for table_name, filename in DATASETS.items():
    file_path = RAW_DATA_PATH / filename
    
    if not file_path.exists():
        print(f"WARNING: File not found -> {file_path}")
        continue
    
    data[table_name] = pd.read_csv(file_path, low_memory=False)

print(f"Datasets successfully loaded: {len(data)}")
```

```python
shape_summary = []

for table_name, df in data.items():
    shape_summary.append({
        "table": table_name,
        "rows": df.shape[0],
        "columns": df.shape[1]
    })

shape_df = pd.DataFrame(shape_summary)

display(
    shape_df.sort_values("rows", ascending=False).reset_index(drop=True)
)
```

```python
for table_name, df in data.items():
    print("\n" + "=" * 80)
    print(f"TABLE: {table_name}")
    print("=" * 80)
    
    print("Columns:")
    for column in df.columns:
        print(f" - {column}")
```

```python
dtype_summary = []

for table_name, df in data.items():
    for column in df.columns:
        dtype_summary.append({
            "table": table_name,
            "column": column,
            "dtype": str(df[column].dtype)
        })

dtype_df = pd.DataFrame(dtype_summary)

display(dtype_df)
```

```python
dataset_summary = []

for table_name, df in data.items():
    dataset_summary.append({
        "table": table_name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "total_cells": df.size,
        "duplicate_rows": df.duplicated().sum(),
        "missing_cells": df.isnull().sum().sum()
    })

dataset_summary_df = pd.DataFrame(dataset_summary)

display(dataset_summary_df)
```

```python
PRIMARY_KEYS = {
    "customers": ["customer_id"],
    "geolocation": ["geolocation_zip_code_prefix"],
    "order_items": ["order_id", "order_item_id"],
    "order_payments": ["order_id", "payment_sequential"],
    "order_reviews": ["review_id"],
    "orders": ["order_id"],
    "products": ["product_id"],
    "sellers": ["seller_id"],
    "translation": ["product_category_name"]
}

primary_key_summary = []

for table_name, keys in PRIMARY_KEYS.items():
    if table_name not in data:
        continue
    
    df = data[table_name]
    
    for key in keys:
        if key not in df.columns:
            primary_key_summary.append({
                "table": table_name,
                "key": key,
                "exists": False,
                "unique": None,
                "null_count": None
            })
            continue
        
        primary_key_summary.append({
            "table": table_name,
            "key": key,
            "exists": True,
            "unique": df[key].is_unique,
            "null_count": df[key].isnull().sum()
        })

primary_key_df = pd.DataFrame(primary_key_summary)

display(primary_key_df)
```

```python
RELATIONSHIPS = [
    {
        "child_table": "orders",
        "child_column": "customer_id",
        "parent_table": "customers",
        "parent_column": "customer_id"
    },
    {
        "child_table": "order_items",
        "child_column": "order_id",
        "parent_table": "orders",
        "parent_column": "order_id"
    },
    {
        "child_table": "order_items",
        "child_column": "product_id",
        "parent_table": "products",
        "parent_column": "product_id"
    },
    {
        "child_table": "order_items",
        "child_column": "seller_id",
        "parent_table": "sellers",
        "parent_column": "seller_id"
    },
    {
        "child_table": "order_payments",
        "child_column": "order_id",
        "parent_table": "orders",
        "parent_column": "order_id"
    },
    {
        "child_table": "order_reviews",
        "child_column": "order_id",
        "parent_table": "orders",
        "parent_column": "order_id"
    },
    {
        "child_table": "products",
        "child_column": "product_category_name",
        "parent_table": "translation",
        "parent_column": "product_category_name"
    }
]

relationships_df = pd.DataFrame(RELATIONSHIPS)

display(relationships_df)
```

```python
relationship_checks = []

for relationship in RELATIONSHIPS:
    
    child_table = relationship["child_table"]
    child_column = relationship["child_column"]
    parent_table = relationship["parent_table"]
    parent_column = relationship["parent_column"]
    
    child_exists = (
        child_table in data
        and child_column in data[child_table].columns
    )
    
    parent_exists = (
        parent_table in data
        and parent_column in data[parent_table].columns
    )
    
    relationship_checks.append({
        "child_table": child_table,
        "child_column": child_column,
        "parent_table": parent_table,
        "parent_column": parent_column,
        "child_column_exists": child_exists,
        "parent_column_exists": parent_exists
    })

relationship_check_df = pd.DataFrame(relationship_checks)

display(relationship_check_df)
```

```python
data_dictionary = []

for table_name, df in data.items():
    for column in df.columns:
        data_dictionary.append({
            "table": table_name,
            "column": column,
            "data_type": str(df[column].dtype),
            "non_null_count": df[column].notna().sum(),
            "null_count": df[column].isna().sum(),
            "unique_values": df[column].nunique(dropna=True),
            "example_value": df[column].dropna().iloc[0]
                if df[column].notna().any()
                else None
        })

data_dictionary_df = pd.DataFrame(data_dictionary)

display(data_dictionary_df)
```

```python
OUTPUT_PATH = PROJECT_ROOT / "data" / "raw"

data_dictionary_file = OUTPUT_PATH / "data_dictionary.csv"

data_dictionary_df.to_csv(
    data_dictionary_file,
    index=False
)

print(f"Data dictionary saved to: {data_dictionary_file}")
```
