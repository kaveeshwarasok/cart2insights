```python
import mysql.connector
import pandas as pd
import numpy as np
from pathlib import Path
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="qwerty",
    database="cart2insights"
)

cursor = conn.cursor()

print("✅ Connected to MySQL successfully")
```

```python
orders = pd.read_sql(
    "SELECT * FROM orders",
    conn
)

print("Orders shape:", orders.shape)
orders.head()
```

```python
order_items = pd.read_sql(
    "SELECT * FROM order_items",
    conn
)

print("Order items shape:", order_items.shape)
order_items.head()
```

```python
customers = pd.read_sql(
    "SELECT * FROM customers",
    conn
)

print("Customers shape:", customers.shape)
customers.head()
```

```python
sellers = pd.read_sql(
    "SELECT * FROM sellers",
    conn
)

print("Sellers shape:", sellers.shape)
sellers.head()
```

```python
payments = pd.read_sql(
    "SELECT * FROM order_payments",
    conn
)

print("Payments shape:", payments.shape)
payments.head()
```

```python
print("=" * 60)
print("FEATURE ENGINEERING DATA CHECK")
print("=" * 60)

print("Orders       :", orders.shape)
print("Order Items  :", order_items.shape)
print("Customers    :", customers.shape)
print("Sellers      :", sellers.shape)
print("Payments     :", payments.shape)

print("=" * 60)

if (
    len(orders) > 0
    and len(order_items) > 0
    and len(customers) > 0
    and len(sellers) > 0
):
    print("✅ Data is available for feature engineering")
else:
    print("❌ One or more required tables are empty")
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    orders[column] = pd.to_datetime(
        orders[column],
        errors="coerce"
    )

print("✅ Date columns converted")
```

```python
order_value = (
    order_items
    .groupby("order_id", as_index=False)
    .agg(
        total_order_value=("price", "sum"),
        total_freight_value=("freight_value", "sum")
    )
)

order_value["total_order_value"] = (
    order_value["total_order_value"]
    + order_value["total_freight_value"]
)

order_value = order_value[
    [
        "order_id",
        "total_order_value"
    ]
]

print("✅ Total order value calculated")

order_value.head()
```

```python
orders = orders.merge(
    order_value,
    on="order_id",
    how="left"
)

orders["total_order_value"] = (
    orders["total_order_value"]
    .fillna(0)
)

print("✅ Total order value added to orders")

orders[
    [
        "order_id",
        "total_order_value"
    ]
].head()
```

```python
orders["delivery_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 60 * 60)

print("✅ Delivery days calculated")

orders[
    [
        "order_id",
        "delivery_days"
    ]
].head()
```

```python
orders["delivery_delay"] = (
    orders["order_delivered_customer_date"]
    - orders["order_estimated_delivery_date"]
).dt.total_seconds() / (24 * 60 * 60)

print("✅ Delivery delay calculated")

orders[
    [
        "order_id",
        "delivery_delay"
    ]
].head()
```

```python
orders["is_delayed"] = np.where(
    orders["delivery_delay"] > 0,
    1,
    0
)

print("✅ Delayed delivery indicator created")

orders[
    [
        "order_id",
        "delivery_delay",
        "is_delayed"
    ]
].head()
```

```python
customer_order_count = (
    orders
    .groupby("customer_id")
    .agg(
        customer_order_count=("order_id", "nunique")
    )
    .reset_index()
)

print("✅ Customer order count calculated")

customer_order_count.head()
```

```python
customer_spending = (
    orders
    .groupby("customer_id")
    .agg(
        customer_total_spending=(
            "total_order_value",
            "sum"
        )
    )
    .reset_index()
)

print("✅ Customer total spending calculated")

customer_spending.head()
```

```python
customer_spending = (
    orders
    .groupby("customer_id")
    .agg(
        customer_total_spending=(
            "total_order_value",
            "sum"
        )
    )
    .reset_index()
)

print("✅ Customer total spending calculated")

customer_spending.head()
```

```python
customer_aov = (
    orders
    .groupby("customer_id")
    .agg(
        average_order_value=(
            "total_order_value",
            "mean"
        )
    )
    .reset_index()
)

print("✅ Customer average order value calculated")

customer_aov.head()
```

```python
repeat_customers = (
    customer_order_count
    .copy()
)

repeat_customers["repeat_customer"] = np.where(
    repeat_customers["customer_order_count"] > 1,
    1,
    0
)

print("✅ Repeat customer indicator created")

repeat_customers.head()
```

```python
seller_revenue = (
    order_items
    .groupby("seller_id")
    .agg(
        seller_revenue=("price", "sum")
    )
    .reset_index()
)

print("✅ Seller revenue calculated")

seller_revenue.head()
```

```python
seller_order_count = (
    order_items
    .groupby("seller_id")
    .agg(
        seller_order_count=("order_id", "nunique")
    )
    .reset_index()
)

print("✅ Seller order count calculated")

seller_order_count.head()
```

```python
seller_features = sellers.merge(
    seller_revenue,
    on="seller_id",
    how="left"
)

seller_features = seller_features.merge(
    seller_order_count,
    on="seller_id",
    how="left"
)

seller_features["seller_revenue"] = (
    seller_features["seller_revenue"]
    .fillna(0)
)

seller_features["seller_order_count"] = (
    seller_features["seller_order_count"]
    .fillna(0)
)

print("✅ Seller feature dataset created")

seller_features.head()
```

```python
customer_features = customers.merge(
    customer_order_count,
    on="customer_id",
    how="left"
)

customer_features = customer_features.merge(
    customer_spending,
    on="customer_id",
    how="left"
)

customer_features = customer_features.merge(
    customer_aov,
    on="customer_id",
    how="left"
)

customer_features = customer_features.merge(
    repeat_customers[
        [
            "customer_id",
            "repeat_customer"
        ]
    ],
    on="customer_id",
    how="left"
)

customer_features["customer_order_count"] = (
    customer_features["customer_order_count"]
    .fillna(0)
)

customer_features["customer_total_spending"] = (
    customer_features["customer_total_spending"]
    .fillna(0)
)

customer_features["average_order_value"] = (
    customer_features["average_order_value"]
    .fillna(0)
)

customer_features["repeat_customer"] = (
    customer_features["repeat_customer"]
    .fillna(0)
)

print("✅ Customer feature dataset created")

customer_features.head()
```

```python
order_features = orders[
    [
        "order_id",
        "customer_id",
        "order_status",
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "total_order_value",
        "delivery_days",
        "delivery_delay",
        "is_delayed"
    ]
].copy()

print("✅ Order feature dataset created")

order_features.head()
```

```python
order_features = order_features.merge(
    customer_features[
        [
            "customer_id",
            "customer_order_count",
            "customer_total_spending",
            "average_order_value",
            "repeat_customer"
        ]
    ],
    on="customer_id",
    how="left"
)

print("✅ Customer features added to orders")

order_features.head()
```

```python
order_item_features = order_item_features.merge(
    seller_features[
        [
            "seller_id",
            "seller_revenue",
            "seller_order_count"
        ]
    ],
    on="seller_id",
    how="left"
)

print("✅ Seller features added to order items")

order_item_features.head()
```

```python
numeric_columns = [
    "total_order_value",
    "delivery_days",
    "delivery_delay",
    "customer_order_count",
    "customer_total_spending",
    "average_order_value",
    "repeat_customer",
    "is_delayed"
]

for column in numeric_columns:
    if column in order_features.columns:
        order_features[column] = (
            order_features[column]
            .fillna(0)
        )

numeric_columns_items = [
    "price",
    "freight_value",
    "item_total_value",
    "seller_revenue",
    "seller_order_count"
]

for column in numeric_columns_items:
    if column in order_item_features.columns:
        order_item_features[column] = (
            order_item_features[column]
            .fillna(0)
        )

print("✅ Missing numeric values handled")
```

```python
print("=" * 70)
print("FEATURE ENGINEERING SUMMARY")
print("=" * 70)

print("\nOrder Features:")
print(order_features.shape)

print("\nCustomer Features:")
print(customer_features.shape)

print("\nSeller Features:")
print(seller_features.shape)

print("\nOrder Item Features:")
print(order_item_features.shape)

print("=" * 70)

print("\nCreated features:")

features_created = [
    "Total order value",
    "Delivery days",
    "Delivery delay",
    "Delayed delivery indicator",
    "Customer order count",
    "Customer total spending",
    "Average order value",
    "Repeat customer indicator",
    "Seller revenue",
    "Seller order count"
]

for feature in features_created:
    print("✅", feature)
```

```python
feature_path = "feature_engineered_data"

Path(feature_path).mkdir(
    exist_ok=True
)

order_features.to_csv(
    f"{feature_path}/order_features.csv",
    index=False
)

customer_features.to_csv(
    f"{feature_path}/customer_features.csv",
    index=False
)

seller_features.to_csv(
    f"{feature_path}/seller_features.csv",
    index=False
)

order_item_features.to_csv(
    f"{feature_path}/order_item_features.csv",
    index=False
)

print("✅ Feature datasets saved successfully")
```
