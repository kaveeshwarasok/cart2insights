# Cart2Insights – Business Insights & Recommendations

This notebook converts the supplied EDA and statistical-analysis outputs into business insights using the required **Observation → Interpretation → Business Impact → Recommendation** structure.

**Source files used:** `monthly_sales.csv`, `category_sales.csv`, `seller_performance.csv`, `state_orders.csv`, `customer_orders.csv`, `delivery_status_review.csv`, and `statistical_analysis_results.csv`.


## 1. Executive Summary

- Total revenue represented in the monthly sales output is **R$ 15,555,072.55** across **99,163 orders**.
- Overall average order value calculated from the monthly totals is **R$ 156.86**.
- The highest monthly revenue occurs in **2017-11**, with **R$ 1,156,317.23**.
- **beleza_saude** is the highest-revenue product category in the supplied category output, generating **R$ 1,255,951.90**.
- **SP** has the highest order volume and revenue among states in the supplied state output: **41,731 orders** and **R$ 5,832,680.88** revenue.
- Delayed orders have an average review score of **2.56**, compared with **4.30** for on-time orders, a gap of **1.73 points**.
- The supplied statistical results reject the null hypothesis for all three requested tests at α = 0.05: t-test, ANOVA, and chi-square.


## 2. Sales Performance Insight

### Observation
Monthly revenue rises substantially through the observed period. The highest reported month is **2017-11**, with revenue of **R$ 1,156,317.23** and **7,521 orders**.

### Interpretation
The monthly output shows a strong increase in transaction activity compared with the early months in the dataset. This indicates that sales volume became substantially larger as the marketplace matured.

### Business Impact
Periods with high order volume can create greater demand on inventory, seller capacity, payment processing, and delivery operations.

### Recommendation
Use monthly sales trends for capacity planning. Increase inventory and seller readiness ahead of historically strong periods and closely monitor fulfillment capacity during peak months.


## 3. Product Category Insight

### Observation
The highest-revenue category is **beleza_saude**, with **R$ 1,255,951.90** revenue and **9,641 units sold**. The category with the highest unit volume is **cama_mesa_banho**, with **11,087 units**.

### Interpretation
Revenue leadership and unit-volume leadership are not necessarily the same. This means category performance should be assessed using both sales value and quantity sold rather than a single metric.

### Business Impact
High-revenue categories can have a larger effect on marketplace revenue, while high-volume categories may have greater inventory and fulfillment requirements.

### Recommendation
Prioritize inventory availability and promotional planning for high-revenue categories, while separately monitoring high-unit categories for stock and operational capacity.


## 4. Geographic Sales Insight

### Observation
**SP** has the highest order volume in the supplied state output, with **41,731 orders**. It also has the highest reported state revenue at **R$ 5,832,680.88**.

### Interpretation
Sales activity is geographically concentrated, with a substantial share of orders coming from the leading state.

### Business Impact
Geographic concentration can influence warehouse planning, seller coverage, delivery capacity, and marketing allocation.

### Recommendation
Use state-level sales data to prioritize logistics capacity and targeted commercial campaigns. Compare lower-volume states separately to identify expansion opportunities.


## 5. Customer Insight

### Observation
The customer output contains **95,828 unique customers**. **92,840 (96.88%)** are one-time customers, while **2,988 (3.12%)** placed more than one order.

### Interpretation
The customer base contains both one-time and repeat purchasing behavior. The repeat-customer share provides a direct measure of how frequently customers return to the marketplace.

### Business Impact
A high proportion of one-time customers creates an opportunity for retention, while repeat customers provide an existing base for cross-selling and repeat-purchase campaigns.

### Recommendation
Create retention campaigns for one-time customers using personalized follow-ups, category recommendations, and incentives. Analyze repeat customers separately to identify products and categories associated with repeat purchasing.


## 6. Seller Performance Insight

### Observation
The seller output contains **3,026 sellers**. The highest-revenue seller in the supplied data generated **R$ 229,302.63** across **1,131 orders** and **1,155 items sold**.

### Interpretation
Seller contribution varies considerably, so marketplace performance is not evenly distributed across sellers.

### Business Impact
High-contribution sellers can have a meaningful effect on marketplace revenue and order fulfillment. Operational problems involving high-volume sellers can therefore affect a large number of orders.

### Recommendation
Track seller revenue, order volume, fulfillment performance, and customer ratings together. Provide targeted operational support to high-volume sellers and investigate recurring delivery or quality issues.


## 7. Delivery & Customer Experience Insight

### Observation
The supplied delivery analysis contains **7,604 delayed orders** with an average review score of **2.56**, versus **87,740 on-time orders** with an average review score of **4.30**. The difference is **1.73 review-score points**.

### Interpretation
In this dataset, delayed and on-time orders are associated with materially different average review scores. The statistical t-test also reports a statistically significant difference at α = 0.05.

### Business Impact
Delivery delays are associated with lower observed customer ratings, making delivery performance an important customer-experience metric.

### Recommendation
Monitor delayed orders by seller and location, identify recurring delivery bottlenecks, and use early-warning monitoring for orders approaching their expected delivery date.


## 8. Statistical Test Insights

### 8.1 Independent Two-Sample T-Test
- **Question:** Delivery status vs review score
- **Test statistic:** -85.000388
- **p-value:** 0.000000e+00
- **Decision:** Reject H0 at α = 0.05.
- **Business meaning:** The supplied results support a statistically significant difference in average review scores between delayed and on-time orders.

### 8.2 One-Way ANOVA
- **Question:** Product category vs order value
- **F-statistic:** 167.966218
- **p-value:** 0.000000e+00
- **Decision:** Reject H0 at α = 0.05.
- **Business meaning:** Average order value differs significantly across at least some product categories. The ANOVA result alone does not identify which category pairs differ; a post-hoc test would be required for that.

### 8.3 Chi-Square Test
- **Question:** Payment method vs order status
- **Chi-square statistic:** 677.968660
- **p-value:** 7.869749e-125
- **Decision:** Reject H0 at α = 0.05.
- **Business meaning:** The supplied results indicate a statistically significant association between payment method and order status. Association does not by itself establish causation.


## 9. Consolidated Recommendations

| Area | Recommendation |
|---|---|
| Sales | Use monthly revenue and order trends for inventory, seller, and fulfillment capacity planning. |
| Categories | Monitor revenue and unit volume separately to support assortment and inventory decisions. |
| Geography | Prioritize logistics and commercial planning in high-volume states while investigating lower-volume regions for expansion. |
| Customers | Build retention campaigns for one-time customers and personalized cross-sell strategies for repeat customers. |
| Sellers | Monitor high-volume sellers using revenue, orders, items sold, delivery performance, and customer feedback together. |
| Delivery | Identify recurring delay patterns by seller and location and intervene before expected delivery dates are missed. |
| Customer Experience | Treat delivery performance as a key customer-experience KPI because delayed orders have substantially lower observed review scores. |
| Statistical Analysis | Use post-hoc analysis after ANOVA if category-pair differences need to be identified, and investigate the practical meaning of the payment-method/order-status association. |


## 10. Important Interpretation Notes

- These insights are based only on the supplied CSV outputs.
- Statistical significance does not by itself prove causation or business importance.
- The ANOVA result establishes that at least some category means differ; it does not identify the specific categories responsible for the difference.
- The chi-square result establishes an association between payment method and order status in the supplied analysis; it does not establish that payment method causes order status.
- The supplied statistical CSV displays p-values as `0.000000e+00` for the t-test and ANOVA. This should be interpreted as an extremely small computed p-value / numerical display underflow rather than literal mathematical zero.
