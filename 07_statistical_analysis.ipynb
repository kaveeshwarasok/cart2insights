TEST 1 — INDEPENDENT TWO-SAMPLE T-TEST
Business question

Do delayed orders receive significantly lower review scores compared to orders delivered on time?

```python
#Prepare order value and delivery data
date_columns = [
    "order_purchase_timestamp",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    orders[column] = pd.to_datetime(
        orders[column],
        errors="coerce"
    )

orders["delivery_delay"] = (
    orders["order_delivered_customer_date"]
    - orders["order_estimated_delivery_date"]
).dt.total_seconds() / (24 * 60 * 60)

orders["delivery_status"] = np.where(
    orders["delivery_delay"] > 0,
    "Delayed",
    "On Time"
)

print(
    orders[
        [
            "order_id",
            "delivery_delay",
            "delivery_status"
        ]
    ].head()
)
```

```python
#Merge delivery status with review score
t_test_data = orders[
    [
        "order_id",
        "delivery_status",
        "delivery_delay"
    ]
].merge(
    reviews[
        [
            "order_id",
            "review_score"
        ]
    ],
    on="order_id",
    how="inner"
)

t_test_data = t_test_data.dropna(
    subset=[
        "delivery_status",
        "review_score"
    ]
)

print("T-test dataset:", t_test_data.shape)

t_test_data.head()
```

```python
group_counts = (
    t_test_data["delivery_status"]
    .value_counts()
)

print("Sample sizes:")
print(group_counts)
```

```python
delayed_reviews = (
    t_test_data.loc[
        t_test_data["delivery_status"] == "Delayed",
        "review_score"
    ]
)

on_time_reviews = (
    t_test_data.loc[
        t_test_data["delivery_status"] == "On Time",
        "review_score"
    ]
)

print(
    "Delayed orders:",
    len(delayed_reviews)
)

print(
    "On-time orders:",
    len(on_time_reviews)
)
```

```python
t_test_summary = pd.DataFrame({
    "Group": [
        "Delayed",
        "On Time"
    ],
    "Sample Size": [
        len(delayed_reviews),
        len(on_time_reviews)
    ],
    "Mean Review Score": [
        delayed_reviews.mean(),
        on_time_reviews.mean()
    ],
    "Std Dev": [
        delayed_reviews.std(),
        on_time_reviews.std()
    ]
})

print(t_test_summary)
```

```python
print("HYPOTHESES — INDEPENDENT TWO-SAMPLE T-TEST")
print("=" * 60)

print(
    "H0: The mean review score is the same for "
    "delayed and on-time orders."
)

print(
    "H1: The mean review score is different for "
    "delayed and on-time orders."
)

print("\nSignificance level (alpha): 0.05")
```

```python
delayed_shapiro = stats.shapiro(
    delayed_reviews.sample(
        min(len(delayed_reviews), 5000),
        random_state=42
    )
)

on_time_shapiro = stats.shapiro(
    on_time_reviews.sample(
        min(len(on_time_reviews), 5000),
        random_state=42
    )
)

print("Shapiro-Wilk Test")
print("=" * 60)

print(
    f"Delayed orders  : "
    f"Statistic={delayed_shapiro.statistic:.4f}, "
    f"p-value={delayed_shapiro.pvalue:.6f}"
)

print(
    f"On-time orders  : "
    f"Statistic={on_time_shapiro.statistic:.4f}, "
    f"p-value={on_time_shapiro.pvalue:.6f}"
)

print(
    "\nNote: With very large samples, "
    "normality tests can be highly sensitive."
)
```

```python
levene_test = stats.levene(
    delayed_reviews,
    on_time_reviews
)

print("Levene's Test for Equality of Variances")
print("=" * 60)

print(
    f"Statistic: {levene_test.statistic:.4f}"
)

print(
    f"p-value : {levene_test.pvalue:.6f}"
)
```

```python
t_statistic, p_value = stats.ttest_ind(
    delayed_reviews,
    on_time_reviews,
    equal_var=False,
    nan_policy="omit"
)

print("INDEPENDENT TWO-SAMPLE T-TEST")
print("=" * 60)

print(
    f"T-statistic: {t_statistic:.4f}"
)

print(
    f"P-value    : {p_value:.10f}"
)
```

```python
alpha = 0.05

print("T-TEST DECISION")
print("=" * 60)

if p_value < alpha:
    print(
        "Decision: Reject H0"
    )
    print(
        "There is statistically significant evidence "
        "of a difference in mean review scores between "
        "delayed and on-time orders."
    )
else:
    print(
        "Decision: Fail to reject H0"
    )
    print(
        "There is not enough statistical evidence "
        "of a difference in mean review scores between "
        "delayed and on-time orders."
    )
```

```python
delayed_mean = delayed_reviews.mean()
on_time_mean = on_time_reviews.mean()

mean_difference = (
    delayed_mean - on_time_mean
)

print("T-TEST BUSINESS SUMMARY")
print("=" * 60)

print(
    f"Delayed orders mean review score : "
    f"{delayed_mean:.3f}"
)

print(
    f"On-time orders mean review score : "
    f"{on_time_mean:.3f}"
)

print(
    f"Mean difference                  : "
    f"{mean_difference:.3f}"
)

print(
    f"P-value                         : "
    f"{p_value:.10f}"
)
```

```python
TEST 2 — ONE-WAY ANOVA
Business question

Does the average order value differ significantly across different product categories?
```

```python
order_value = (
    order_items
    .groupby("order_id", as_index=False)
    .agg(
        total_order_value=("price", "sum")
    )
)

anova_data = order_items[
    [
        "order_id",
        "product_id"
    ]
].merge(
    products[
        [
            "product_id",
            "product_category_name"
        ]
    ],
    on="product_id",
    how="left"
)

anova_data = anova_data.merge(
    order_value,
    on="order_id",
    how="left"
)

anova_data = anova_data.dropna(
    subset=[
        "product_category_name",
        "total_order_value"
    ]
)

print("ANOVA dataset:", anova_data.shape)

anova_data.head()
```

```python
category_counts = (
    anova_data["product_category_name"]
    .value_counts()
)

print("Product category sample sizes:")
print(category_counts.head(20))
```

```python
# Use categories with at least 30 observations
valid_categories = (
    category_counts[
        category_counts >= 30
    ]
    .index
)

anova_data_filtered = anova_data[
    anova_data["product_category_name"]
    .isin(valid_categories)
].copy()

print(
    "Categories included:",
    len(valid_categories)
)

print(
    "Observations included:",
    len(anova_data_filtered)
)
```

```python
anova_summary = (
    anova_data_filtered
    .groupby("product_category_name")
    .agg(
        sample_size=(
            "total_order_value",
            "count"
        ),
        mean_order_value=(
            "total_order_value",
            "mean"
        ),
        std_order_value=(
            "total_order_value",
            "std"
        )
    )
    .sort_values(
        "mean_order_value",
        ascending=False
    )
)

print(
    anova_summary.head(20)
)
```

```python
print("HYPOTHESES — ONE-WAY ANOVA")
print("=" * 60)

print(
    "H0: The mean order value is the same "
    "across all product categories."
)

print(
    "H1: At least one product category has "
    "a different mean order value."
)

print("\nSignificance level (alpha): 0.05")
```

```python
anova_groups = []

for category in valid_categories:

    values = (
        anova_data_filtered.loc[
            anova_data_filtered["product_category_name"]
            == category,
            "total_order_value"
        ]
        .dropna()
    )

    if len(values) >= 30:
        anova_groups.append(values)

print(
    "Number of groups:",
    len(anova_groups)
)
```

```python
levene_anova = stats.levene(
    *anova_groups
)

print("Levene's Test for Homogeneity of Variance")
print("=" * 60)

print(
    f"Statistic: {levene_anova.statistic:.4f}"
)

print(
    f"P-value  : {levene_anova.pvalue:.10f}"
)
```

```python
anova_statistic, anova_p_value = stats.f_oneway(
    *anova_groups
)

print("ONE-WAY ANOVA")
print("=" * 60)

print(
    f"F-statistic: {anova_statistic:.4f}"
)

print(
    f"P-value   : {anova_p_value:.10f}"
)
```

```python
alpha = 0.05

print("ANOVA DECISION")
print("=" * 60)

if anova_p_value < alpha:
    print(
        "Decision: Reject H0"
    )
    print(
        "There is statistically significant evidence "
        "that average order value differs across "
        "at least some product categories."
    )
else:
    print(
        "Decision: Fail to reject H0"
    )
    print(
        "There is not enough statistical evidence "
        "that average order value differs across "
        "product categories."
    )
```

```python
top_anova_categories = (
    anova_summary
    .head(10)
    .index
)

plot_data = anova_data_filtered[
    anova_data_filtered["product_category_name"]
    .isin(top_anova_categories)
]

plt.figure(figsize=(14, 7))

sns.boxplot(
    data=plot_data,
    x="product_category_name",
    y="total_order_value"
)

plt.title(
    "Order Value Distribution Across Top Product Categories"
)

plt.xlabel("Product Category")
plt.ylabel("Total Order Value")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()
plt.show()
```

TEST 3 — CHI-SQUARE TEST
Business question

Is there a significant association between payment method and order status?

```python
chi_data = payments[
    [
        "order_id",
        "payment_type"
    ]
].merge(
    orders[
        [
            "order_id",
            "order_status"
        ]
    ],
    on="order_id",
    how="inner"
)

chi_data = chi_data.dropna(
    subset=[
        "payment_type",
        "order_status"
    ]
)

print(
    "Chi-Square dataset:",
    chi_data.shape
)

chi_data.head()
```

```python
contingency_table = pd.crosstab(
    chi_data["payment_type"],
    chi_data["order_status"]
)

print("CONTINGENCY TABLE")
print("=" * 70)

print(contingency_table)
```

```python
print("HYPOTHESES — CHI-SQUARE TEST OF INDEPENDENCE")
print("=" * 60)

print(
    "H0: Payment method and order status are independent."
)

print(
    "H1: Payment method and order status are associated."
)

print("\nSignificance level (alpha): 0.05")
```

```python
chi2_stat, chi2_p, degrees_freedom, expected = (
    stats.chi2_contingency(
        contingency_table
    )
)

expected_table = pd.DataFrame(
    expected,
    index=contingency_table.index,
    columns=contingency_table.columns
)

print("EXPECTED FREQUENCIES")
print("=" * 70)

print(expected_table)
```

```python
expected_values = expected.flatten()

cells_below_5 = (
    (expected_values < 5).sum()
)

percentage_below_5 = (
    cells_below_5
    / len(expected_values)
    * 100
)

print("CHI-SQUARE ASSUMPTION CHECK")
print("=" * 60)

print(
    f"Cells with expected frequency < 5: "
    f"{cells_below_5}"
)

print(
    f"Percentage of cells < 5: "
    f"{percentage_below_5:.2f}%"
)

if percentage_below_5 <= 20:
    print(
        "✅ Expected-frequency assumption is acceptable "
        "under the common rule of thumb."
    )
else:
    print(
        "⚠️ More than 20% of expected cells are below 5. "
        "Interpret the Chi-Square result cautiously."
    )
```

```python
print("CHI-SQUARE TEST OF INDEPENDENCE")
print("=" * 60)

print(
    f"Chi-Square statistic: {chi2_stat:.4f}"
)

print(
    f"Degrees of freedom  : {degrees_freedom}"
)

print(
    f"P-value             : {chi2_p:.10f}"
)
```

```python
alpha = 0.05

print("CHI-SQUARE DECISION")
print("=" * 60)

if chi2_p < alpha:
    print(
        "Decision: Reject H0"
    )
    print(
        "There is statistically significant evidence "
        "of an association between payment method "
        "and order status."
    )
else:
    print(
        "Decision: Fail to reject H0"
    )
    print(
        "There is not enough statistical evidence "
        "of an association between payment method "
        "and order status."
    )
```
