```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("✅ Libraries imported")
```

```python
import mysql.connector

conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="qwerty",
    database="cart2insights"
)

cursor = conn.cursor()

print("✅ Connected to MySQL successfully")
```

```python
orders = pd.read_sql(
    """
    SELECT *
    FROM orders
    """,
    conn
)

print("Orders:", orders.shape)
orders.head()
```

```python
order_items = pd.read_sql(
    """
    SELECT *
    FROM order_items
    """,
    conn
)

print("Order items:", order_items.shape)
order_items.head()
```

```python
customers = pd.read_sql(
    """
    SELECT *
    FROM customers
    """,
    conn
)

print("Customers:", customers.shape)
customers.head()
```

```python
sellers = pd.read_sql(
    """
    SELECT *
    FROM sellers
    """,
    conn
)

print("Sellers:", sellers.shape)
sellers.head()
```

```python
products = pd.read_sql(
    """
    SELECT *
    FROM products
    """,
    conn
)

print("Products:", products.shape)
products.head()
```

```python
payments = pd.read_sql(
    """
    SELECT *
    FROM order_payments
    """,
    conn
)

print("Payments:", payments.shape)
payments.head()
```

```python
reviews = pd.read_sql(
    """
    SELECT *
    FROM order_reviews
    """,
    conn
)

print("Reviews:", reviews.shape)
reviews.head()
```

```python
geolocation = pd.read_sql(
    """
    SELECT *
    FROM geolocation
    """,
    conn
)

print("Geolocation:", geolocation.shape)
geolocation.head()
```

```python
print("=" * 70)
print("EDA DATASET CHECK")
print("=" * 70)

datasets = {
    "Orders": orders,
    "Order Items": order_items,
    "Customers": customers,
    "Sellers": sellers,
    "Products": products,
    "Payments": payments,
    "Reviews": reviews,
    "Geolocation": geolocation
}

for name, df in datasets.items():
    print(f"{name:20} : {df.shape[0]:,} rows | {df.shape[1]} columns")

print("=" * 70)
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    orders[column] = pd.to_datetime(
        orders[column],
        errors="coerce"
    )

reviews["review_creation_date"] = pd.to_datetime(
    reviews["review_creation_date"],
    errors="coerce"
)

reviews["review_answer_timestamp"] = pd.to_datetime(
    reviews["review_answer_timestamp"],
    errors="coerce"
)

print("✅ Date columns converted")
```

```python
order_value = (
    order_items
    .groupby("order_id", as_index=False)
    .agg(
        product_value=("price", "sum"),
        freight_value=("freight_value", "sum")
    )
)

order_value["total_order_value"] = (
    order_value["product_value"]
    + order_value["freight_value"]
)

print("✅ Total order value calculated")

order_value.head()
```

```python
orders = orders.merge(
    order_value[
        [
            "order_id",
            "total_order_value"
        ]
    ],
    on="order_id",
    how="left"
)

orders["total_order_value"] = (
    orders["total_order_value"]
    .fillna(0)
)

print("✅ Order value added")

orders[
    [
        "order_id",
        "total_order_value"
    ]
].head()
```

```python
orders["delivery_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 60 * 60)

orders["delivery_delay"] = (
    orders["order_delivered_customer_date"]
    - orders["order_estimated_delivery_date"]
).dt.total_seconds() / (24 * 60 * 60)

orders["is_delayed"] = np.where(
    orders["delivery_delay"] > 0,
    1,
    0
)

print("✅ Delivery features created")

orders[
    [
        "order_id",
        "delivery_days",
        "delivery_delay",
        "is_delayed"
    ]
].head()
```

```python
monthly_sales = (
    orders
    .dropna(subset=["order_purchase_timestamp"])
    .assign(
        month=lambda x:
        x["order_purchase_timestamp"].dt.to_period("M").astype(str)
    )
    .groupby("month", as_index=False)
    .agg(
        revenue=("total_order_value", "sum"),
        orders=("order_id", "nunique")
    )
)

print("✅ Monthly sales dataset created")

monthly_sales.head()
```

```python
aov_by_status = (
    orders
    .groupby("order_status", as_index=False)
    .agg(
        average_order_value=(
            "total_order_value",
            "mean"
        )
    )
    .sort_values(
        "average_order_value",
        ascending=False
    )
)

print(aov_by_status)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=aov_by_status,
    x="order_status",
    y="average_order_value"
)

plt.title("Average Order Value by Order Status")
plt.xlabel("Order Status")
plt.ylabel("Average Order Value")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
delivery_review = orders[
    [
        "order_id",
        "delivery_delay"
    ]
].merge(
    reviews[
        [
            "order_id",
            "review_score"
        ]
    ],
    on="order_id",
    how="inner"
)

delivery_review = delivery_review.dropna(
    subset=[
        "delivery_delay",
        "review_score"
    ]
)

print(delivery_review.head())

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=delivery_review,
    x="review_score",
    y="delivery_delay"
)

plt.title("Delivery Delay vs Review Score")
plt.xlabel("Review Score")
plt.ylabel("Delivery Delay (Days)")
plt.tight_layout()
plt.show()
```

```python
delivery_status_review = (
    delivery_review
    .assign(
        delivery_status=np.where(
            delivery_review["delivery_delay"] > 0,
            "Delayed",
            "On Time"
        )
    )
    .groupby("delivery_status", as_index=False)
    .agg(
        average_review_score=(
            "review_score",
            "mean"
        ),
        review_count=("review_score", "count")
    )
)

print(delivery_status_review)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=delivery_status_review,
    x="delivery_status",
    y="average_review_score"
)

plt.title("Average Review Score by Delivery Status")
plt.xlabel("Delivery Status")
plt.ylabel("Average Review Score")
plt.tight_layout()
plt.show()
```

```python
category_sales = (
    order_items
    .merge(
        products[
            [
                "product_id",
                "product_category_name"
            ]
        ],
        on="product_id",
        how="left"
    )
    .groupby(
        "product_category_name",
        as_index=False
    )
    .agg(
        revenue=("price", "sum"),
        units_sold=("product_id", "count")
    )
    .sort_values(
        "revenue",
        ascending=False
    )
)

print(category_sales.head(10))
```

```python
top_categories = (
    category_sales
    .head(10)
    .sort_values(
        "revenue",
        ascending=True
    )
)

plt.figure(figsize=(10, 7))

sns.barplot(
    data=top_categories,
    x="revenue",
    y="product_category_name"
)

plt.title("Top 10 Product Categories by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Product Category")
plt.tight_layout()
plt.show()
```

```python
seller_performance = (
    order_items
    .groupby("seller_id", as_index=False)
    .agg(
        revenue=("price", "sum"),
        orders=("order_id", "nunique"),
        items_sold=("order_id", "count")
    )
    .sort_values(
        "revenue",
        ascending=False
    )
)

print(seller_performance.head(10))
```

```python
top_sellers = (
    seller_performance
    .head(10)
    .sort_values(
        "revenue",
        ascending=True
    )
)

plt.figure(figsize=(10, 7))

sns.barplot(
    data=top_sellers,
    x="revenue",
    y="seller_id"
)

plt.title("Top 10 Sellers by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Seller ID")
plt.tight_layout()
plt.show()
```

```python
customer_orders = (
    orders
    .groupby("customer_id", as_index=False)
    .agg(
        order_count=("order_id", "nunique"),
        total_spending=("total_order_value", "sum")
    )
)

print(customer_orders.describe())
```

```python
customer_orders = (
    orders
    .merge(
        customers[
            [
                "customer_id",
                "customer_unique_id"
            ]
        ],
        on="customer_id",
        how="left"
    )
    .groupby(
        "customer_unique_id",
        as_index=False
    )
    .agg(
        order_count=("order_id", "nunique"),
        total_spending=("total_order_value", "sum")
    )
)

print(customer_orders.describe())

customer_orders.head()
```

```python
customer_orders["customer_type"] = np.where(
    customer_orders["order_count"] > 1,
    "Repeat Customer",
    "One-time Customer"
)

customer_type_counts = (
    customer_orders["customer_type"]
    .value_counts()
)

print(customer_type_counts)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=customer_type_counts.index,
    y=customer_type_counts.values
)

plt.title("One-time vs Repeat Customers")
plt.xlabel("Customer Type")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

```python
customer_orders_location = (
    orders
    .merge(
        customers[
            [
                "customer_id",
                "customer_state"
            ]
        ],
        on="customer_id",
        how="left"
    )
)

state_orders = (
    customer_orders_location
    .groupby(
        "customer_state",
        as_index=False
    )
    .agg(
        orders=("order_id", "nunique"),
        revenue=("total_order_value", "sum")
    )
    .sort_values(
        "orders",
        ascending=False
    )
)

print(state_orders.head(10))
```

```python
top_states = (
    state_orders
    .head(10)
    .sort_values(
        "orders",
        ascending=True
    )
)

plt.figure(figsize=(10, 7))

sns.barplot(
    data=top_states,
    x="orders",
    y="customer_state"
)

plt.title("Top 10 States by Order Volume")
plt.xlabel("Number of Orders")
plt.ylabel("State")
plt.tight_layout()
plt.show()
```
